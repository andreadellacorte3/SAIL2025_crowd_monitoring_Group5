```python
import pandas as pd

def station_counts():
    df = pd.read_csv('sensor_mapping_with_coords.csv')
    
    station_sensors = [
    "GASA-01-C",
    "GASA-01-B",
    "GASA-01-A2",
    "GASA-01-A1",
    "GVCV-04",
    "GVCV-03",
    "GVCV-01",
    "GACM-04",
    "GASA-02-02",
    "GASA-02-01",
    "GASA-04",
    "GVCV-05-B",
    "GVCV-05-A"]

    
    target_columns = df[df["sensor_id"].isin(station_sensors)][
      "column_name"
    ].tolist()

    df2 = pd.read_csv('SAIL2025_LVMA_data_3min_20August-25August2025_flow.csv', usecols=lambda col: col == "timestamp" or col in target_columns)
  
    sensor_cols = [c for c in df2.columns if c != "timestamp"]
    df2["pedestrians"] = df2[sensor_cols].sum(axis=1)


    return df2[["timestamp", "pedestrians"]]


df_stazione = station_counts()
print(df_stazione.head())
```
