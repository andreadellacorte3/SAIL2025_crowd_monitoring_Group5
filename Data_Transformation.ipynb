Depending on the day, the hours of the SAIL event vary. We create masks to keep rows that only measure during these hours. When joining the columns we realized that some rows had hours or minutes without 2 digits. For example, 04:05 was 4 in column hour and 5 in column minute. We had to change this for the masks.

```python
SAIL2025_LVMA["hourminute"] = (SAIL2025_LVMA["hour"].astype(str).str.zfill(2) + SAIL2025_LVMA["minute"].astype(str).str.zfill(2))
SAIL2025_LVMA["hourminute"] = SAIL2025_LVMA["hourminute"].astype(int)
```

```python
wen = (SAIL2025_LVMA["day"] == 20) & (SAIL2025_LVMA["hourminute"] >= 1000) & (SAIL2025_LVMA["hourminute"] <= 2245)
thu = (SAIL2025_LVMA["day"] == 21) & (SAIL2025_LVMA["hourminute"] >= 800) & (SAIL2025_LVMA["hourminute"] <= 2245)
fri = (SAIL2025_LVMA["day"] == 22) & (SAIL2025_LVMA["hourminute"] >= 900) & (SAIL2025_LVMA["hourminute"] <= 2245)
sat = (SAIL2025_LVMA["day"] == 23) & (SAIL2025_LVMA["hourminute"] >= 900) & (SAIL2025_LVMA["hourminute"] <= 2245)
sun = (SAIL2025_LVMA["day"] == 24) & (SAIL2025_LVMA["hourminute"] >= 900) & (SAIL2025_LVMA["hourminute"] <= 2100)
```

```python
filtered = SAIL2025_LVMA[wen | thu | fri | sat | sun]
len(filtered.index)
```

Making this filters we get roughly 1000 rows less compared to the original dataframe.

```python
filtered.isna().sum().sum() # No missing values
```

Now check if there are zero value readings in any sensors during this period to detect malfunctions.

```python
for column in filtered.columns:

    temp = filtered[filtered[column] == 0]
    print(f"{column} - {len(temp.index)}")
```
