Depending on the day, the hours of the SAIL event vary. We create masks to keep rows that only measure during these hours. When joining the columns we realized that some rows had hours or minutes without 2 digits. For example, 04:05 was 4 in column hour and 5 in column minute. We had to change this for the masks.

```python
SAIL2025_LVMA["hourminute"] = (SAIL2025_LVMA["hour"].astype(str).str.zfill(2) + SAIL2025_LVMA["minute"].astype(str).str.zfill(2))
SAIL2025_LVMA["hourminute"] = SAIL2025_LVMA["hourminute"].astype(int)
```

```python
wen = (SAIL2025_LVMA["day"] == 20) & (SAIL2025_LVMA["hourminute"] >= 1000) & (SAIL2025_LVMA["hourminute"] <= 2245)
thu = (SAIL2025_LVMA["day"] == 21) & (SAIL2025_LVMA["hourminute"] >= 800) & (SAIL2025_LVMA["hourminute"] <= 2245)
fri = (SAIL2025_LVMA["day"] == 22) & (SAIL2025_LVMA["hourminute"] >= 900) & (SAIL2025_LVMA["hourminute"] <= 2245)
sat = (SAIL2025_LVMA["day"] == 23) & (SAIL2025_LVMA["hourminute"] >= 900) & (SAIL2025_LVMA["hourminute"] <= 2245)
sun = (SAIL2025_LVMA["day"] == 24) & (SAIL2025_LVMA["hourminute"] >= 900) & (SAIL2025_LVMA["hourminute"] <= 2100)
```

```python
filtered = SAIL2025_LVMA[wen | thu | fri | sat | sun]
len(filtered.index)
```

Making this filters we get roughly 1000 rows less compared to the original dataframe.

With this version we can now crete an extra column with the time stamp for each row and drop the row with the timestamp. This will have an associated time and can be used to merge with NWB dataset.

```python
# Detectar las filas que contienen un timestamp ISO
timestamp_pattern = (
    r"^\d{4}-\d{2}-\d{2}T"
    r"\d{2}:\d{2}:\d{2}"
)

tomtom["timestamp"] = tomtom["time"].str.match(
    timestamp_pattern, na=False
)

# Guardar el timestamp solo en las filas donde aparece
tomtom["timestamp"] = tomtom["time"].where(tomtom["timestamp"])

# Propagar el timestamp hacia abajo
tomtom["timestamp"] = tomtom["timestamp"].ffill()

# Eliminar las filas que contienen timestamps
# en lugar de datos de tráfico
tomtom = tomtom[
    ~tomtom["time"].str.match(timestamp_pattern, na=False)
].copy()

# Renombrar las columnas de datos
tomtom = tomtom.rename(columns={
    "time": "id",
    "data": "traffic_level"
})

# Conservar únicamente las columnas necesarias
tomtom = tomtom[["timestamp", "id", "traffic_level"]]

# Limpiar espacios y convertir tipos
tomtom["id"] = pd.to_numeric(tomtom["id"], errors="coerce")
tomtom["traffic_level"] = pd.to_numeric(
    tomtom["traffic_level"], errors="coerce"
)

tomtom = tomtom.dropna(subset=["timestamp", "id", "traffic_level"])
```

```python
tomtom["id"] = tomtom["id"].astype(int)
```
