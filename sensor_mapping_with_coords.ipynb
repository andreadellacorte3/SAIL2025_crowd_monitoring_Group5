```python
import pandas as pd


df_flow = pd.read_csv('SAIL2025_LVMA_data_3min_20August-25August2025_flow.csv')

df_locations = pd.read_excel('sensor-location.xlsx')

print("Excel columns loaded successfully:")
print(df_locations.columns)


time_cols = ['timestamp', 'hour', 'minute', 'day', 'month', 'weekday', 'is_weekend']
sensor_cols = [col for col in df_flow.columns if col not in time_cols]

sensor_mapping = []
for col in sensor_cols:
    if '_' in col:
        parts = col.rsplit('_', 1)
        base_id = parts[0]
        direction = parts[1]
    else:
        base_id = col
        direction = 'N/A'
        
    sensor_mapping.append({
        'column_name': col,
        'sensor_id': base_id,
        'direction': direction
    })

df_mapping = pd.DataFrame(sensor_mapping)

print("\nSensor mapping preview:")
display(df_mapping)
```

```python
import pandas as pd


df_flow = pd.read_csv('SAIL2025_LVMA_data_3min_20August-25August2025_flow.csv')


df_locations = pd.read_excel('sensor-location.xlsx', engine='openpyxl')

print("Excel columns loaded successfully:")
print(df_locations.columns)


time_cols = ['timestamp', 'hour', 'minute', 'day', 'month', 'weekday', 'is_weekend']
sensor_cols = [col for col in df_flow.columns if col not in time_cols]

sensor_mapping = []
for col in sensor_cols:
    if '_' in col:
        parts = col.rsplit('_', 1)
        base_id = parts[0]
        direction = parts[1]
    else:
        base_id = col
        direction = 'N/A'
        
    sensor_mapping.append({
        'column_name': col,
        'sensor_id': base_id,
        'direction': direction
    })

df_mapping = pd.DataFrame(sensor_mapping)

print("\nSensor mapping preview:")
print(df_mapping)


df_mapping.to_csv('sensor_mapping_output.csv', index=False)
print("\nMapping saved to 'sensor_mapping_output.csv' successfully!")
```

```python
import pandas as pd

df_mapping = pd.read_csv('sensor_mapping_output.csv')

df_locations = pd.read_excel('sensor-location.xlsx', engine='openpyxl')


df_locations_clean = pd.DataFrame({
    'sensor_key': df_locations.iloc[:, 0],
    'location_name': df_locations.iloc[:, 1],
    'lat_long': df_locations.iloc[:, 2]
})

df_locations_clean = df_locations_clean.dropna(subset=['sensor_key'])

df_merged = pd.merge(
    df_mapping,
    df_locations_clean,
    left_on='sensor_id',
    right_on='sensor_key',
    how='left'
)

df_merged = df_merged.drop(columns=['sensor_key'])

print("\nMerged mapping with coordinates preview:")
display(df_merged.head(10))

df_merged.to_csv('sensor_mapping_with_coords.csv', index=False)
print("\nFinal dataset saved to 'sensor_mapping_with_coords.csv' successfully!")
```
